## 7.2 Comparison Operators

Comparison operators are tools that allow us to compare two values. The act of comparison can be **applied to variables** of any type, be it integers, strings, lists, etc. However, the **result of this comparison is always a Boolean**: either `True` or `False`.

Here are some common comparison operators:

- `==`: Equal to
- `!=`: Not equal to
- `<`: Less than
- `<=`: Less than **or** equal to
- `>`: Greater than
- `>=`: Greater than **or** equal to


````python
x = 5
y = 10

print(x == y) # False
print(x != y) # True
print(x < y) # True
print(x <= y) # True
print(x > y) # False
print(x >= y) # False
````

Run the code above and then try changing the values and run again. You can think of `x == y` as asking the question: "Is `x` **equal** in value to `y`". The answer must either be a `True` or a`False`. Similarly you can think of `x != y` as asking the question: "Is `x` **not equal** in value to `y`" and so on. For the case `x<=y` we are asking two questions: "Is `x` **equal** in value to `y`" **or** "is `x` **less than** `y`". If either condition is true then the operator returns a `True`. 

Remember: the act of comparison can be applied to any type:

````python
string1 = "hello"
string2 = "world"

print(string1 == string2) # False
````

## 7.3 Introducing Conditional Statements: `if`, `else`, and `elif`

Armed with the power of Booleans and comparisons, we can introduce logic into our programs using conditional statements.

### The `if` Statement

The `if` statement checks a condition and executes a block of code if that condition is `True`. Run the code block below to give it a trial run:

````python
temperature = 15

if temperature < 20:
    print("It's chilly outside!") # This is inside the body of the if
    print("You might need a jacket.") # So is this!
````

Here, since `temperature` is less than `20`, the message "It's chilly outside!" gets printed. We also the message "You might need a jacked printed". Change the value of `temperature` to be 25 now and see what gets printed. Let's break it down: 

- 1. The line `if temperature < 20:` is a condition that checks if the value of `temperature` is less than `20`.

- 2. The subsequent lines that are indented (in this case, two `print()` statements) form the body of the `if` statement. Both of these lines will be executed if the condition is `True`.

- 3. The body of the `if` statement is indicated entirely by its indentation. Every line that's indented by the same amount (or more) after the `if` statement is considered part of the `if` block.

- 4. The end of the `if` block (or the body of the `if` statement) is recognized when the interpreter encounters a line that's indented at the same level as the `if`, or less. This means the block can be as long or as short as you need it to be, but it always ends when the indentation decreases to the level of the `if` or goes even further left.

To further illustrate the point, consider this:

````python
temperature = 15

if temperature < 20:
    print("It's chilly outside!") # Only gets printed if temperature < 20
    print("You might need a jacket.") # Only gets printed if temperature < 20
print("This line is outside of the if block.") # Always gets printed
````

In this example, the last `print()` statement is not indented, so it's not part of the `if` block. It will be executed regardless of the value of `temperature`, whereas the two `print()` statements inside the `if` block will only be executed if the condition is `True`.

The consistent use of indentation ensures that Python code is often cleaner and easier to read than code in many other languages, but it also means that maintaining consistent indentation is critical to ensuring your code behaves as expected.

### The `elif` Statement

What if we have more than two possible actions based on multiple conditions? Enter the `elif` statement, which is short for "else if". Here's a simple example.



````python
temperature = 20

if temperature < 20:
    print("It's chilly outside!")
    print("You should bring a coat!")
elif temperature == 20:
    print("The temperature is just right!")
else:
    print("It's warm outside!")
    print("Wear a t-shirt!")
````

Note: You can have as many `elif`s as you like. In the case above for the value of `temperature` being 20 the message "The temperature is just right!" is printed. You should change the value of `temperature` to confirm the other messages will also be printed depending on which conditions is `True`

### `or` Operator

The `or` operator returns `True` if at least one of the conditions on its sides is true. It only returns `False` if both conditions are false. Here is a simple code example, try and run it:

````python
x = 10
y = 12

if x < 15 or y > 15:
    print("At least one of the conditions is true.")
````

Here, even though `y` is not greater than `15`, the condition `x < 15` is true, so the message is printed. Only one codition need be true.

## 7.6 Order of Operations in Logical Expressions

Lets consider what we meant by the **order of operations** above. 
In the context of logical and arithmetic operators in programming, "order of operations" refers to the sequence in which operators are evaluated. Just as in mathematics where multiplication and division are performed before addition and subtraction, in programming, certain operators take precedence over others.

For logical operators in Python:

1. The `not` operator has the highest precedence.
2. The `and` operator is evaluated next.
3. The `or` operator has the lowest precedence.

This means that in an expression like `not A and B`, the `not` operation will be evaluated first, then the `and` operation.

Consider the following example:

````python
A = False
B = True
C = True

result = A and not B or C
print(result)
````

In this expression, precedence groups the code as:

```python
(A and (not B)) or C
```

With the values above, `A` is `False`, so `A and (not B)` is `False`. Python **short-circuits** this part: it does not need to evaluate `not B`. It then evaluates `C`, which is `True`, so the final result is `True`.

Precedence determines **how an expression is grouped**; short-circuiting determines whether later parts need to be evaluated. Parentheses can make the grouping clearer or change it. For example, `A and ((not B) or C)` groups the same terms differently.
