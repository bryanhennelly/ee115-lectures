## 9.3 Adding a While Loop

Now that we understand how to bisect the interval and adjust the bounds using an `if` statement, it's time to automate this process using a **while loop**.

### Why Use a While Loop?

A `while` loop allows us to repeat the bisecting process until we have narrowed the interval to a point where the bounds are close enough to each other. This means we will keep adjusting the bounds until the difference between the upper and lower bounds is very small (close to zero).

### Choosing a Stopping Condition

To decide when to stop the loop, we need a **tolerance** level, which is the maximum allowable difference between the **upper** and **lower** bounds. For example, if we set a tolerance of 0.0001, the loop will stop when the difference between the upper and lower bounds is less than this value.

This makes our program much more efficient, as it will keep bisecting the interval automatically until it reaches the precision we want.

### Let's write the code

The code below will keep adjusting the bounds using a `while` loop until the difference between the bounds is less than the tolerance level. 


A stopping condition of `guess_high - guess_low > tolerance` means the final width is **at most** the tolerance. If the true answer stays within the interval, its distance from the final midpoint is at most half that width. This controls error in the **square-root estimate**, not directly the difference between its square and the target.

### Guided exercise 11: Stop when it is accurate enough

Write the full loop for `the_square = 20`, with starting bounds 0 and 20. Continue while the width exceeds `0.001`. After the loop, calculate and print the final midpoint and width.

**Hint:** Replace the fixed step limit with a width test. Calculate the final answer from the final bounds, outside the loop.

```python
# Exercise 11: Stop when it is accurate enough

```

```python
the_square = 43
guess_low = 0
guess_high = the_square

tolerance = 0.0001

while (guess_high - guess_low) > tolerance:
    midpoint = (guess_low + guess_high) / 2
    
    if midpoint ** 2 > the_square:
        guess_high = midpoint
    else:
        guess_low = midpoint
    
    print("Current lower bound:", guess_low)
    print("Current upper bound:", guess_high)
    print("Difference:", guess_high - guess_low)

approx_sqrt = (guess_high + guess_low) / 2
print("\nApproximate square root of 43:", approx_sqrt)
```

### Explanation of code so far

1. We set up our initial bounds (`guess_low` and `guess_high`) and our tolerance level.
2. The `while` loop runs as long as the difference between `guess_high` and `guess_low` is greater than the tolerance.
3. Inside the loop:
   - We calculate the midpoint and check if its square is greater or less than 43.
   - Depending on the result, we update either the `guess_high` or `guess_low` bound.
   - We print the current bounds and the difference between them to observe the narrowing process.
4. When the loop ends, we calculate the approximate square root as the midpoint of the final bounds.

This version finds the square root of 43. For other non-negative numbers, we first need suitable starting bounds; the next section explains this. You can try changing the tolerance level to see how it affects the precision of the result!

## 9.4 What About Values Between 0 and 1?

We're not done yet!! When calculating the square root of numbers between 0 and 1 (e.g., 0.25 or 0.5), the algorithm behaves slightly differently. This is because the square of a number between 0 and 1 is **smaller** than the number itself. 

### Understanding the Difference

For example, if we take a number like 0.5:
- Its square is 0.25, which is **less** than 0.5.
- This is the opposite of what happens when we square numbers greater than 1, where the square is always **larger** than the number.

### Adjusting the Bisection Method

For numbers between 0 and 1:
- We still set up the lower and upper bounds, but this time the **upper bound should start at 1**, because the square root of any number between 0 and 1 is never greater than 1.
- The comparison and bound updates stay exactly the same. Only the initial upper bound changes.

### Let's write the code for square root of 0.25

In the code below, we will calculate the square root of 0.25 using a while loop and an `if` statement.

### Guided exercise 15: Solve a value below one

Find the square root of `the_square = 0.09` by bisection. Use bounds 0 and 1 and tolerance 0.0001. Print your estimate.

**Hint:** Keep the same comparison and update rule as before. Only the initial bounds change; the answer should be close to 0.3.

```python
# Exercise 15: Solve a value below one

```

```python
guess_low = 0
guess_high = 1
the_square = 0.25
tolerance = 0.0001

while (guess_high - guess_low) > tolerance:
    midpoint = (guess_low + guess_high) / 2
    
    if midpoint ** 2 > the_square:
        guess_high = midpoint
    else:
        guess_low = midpoint
    
    print("Current lower bound:", guess_low)
    print("Current upper bound:", guess_high)
    print("Difference:", guess_high - guess_low)

approx_sqrt = (guess_high + guess_low) / 2
print("\nApproximate square root of", the_square, ":", approx_sqrt)
```

### Explanation

1. We set up the initial bounds:
   - `guess_low = 0` because the square root of any number between 0 and 1 is always greater than 0.
   - `guess_high = 1` because the square root of any number between 0 and 1 is always less than or equal to 1.
2. We choose a `tolerance` level of 0.0001, which sets how precise we want our result to be.
3. Inside the `while` loop:
   - We calculate the `midpoint` and check if squaring it results in a value greater than or less than the target (`the_square`).
   - If the square of the `midpoint` is greater than the value, it means the square root must be smaller, so we adjust the `guess_high` bound.
   - If the square of the `midpoint` is smaller, it means the square root must be larger, so we adjust the `guess_low` bound.
4. The loop continues until the difference between `guess_high` and `guess_low` is less than our tolerance.
5. We calculate the square root as the average of the final bounds.

### Complete worked example

Try your own program above first, then use this version to compare the structure.

```python
the_square = float(input("Enter a number greater than or equal to 0: "))
while the_square < 0:
    the_square = float(input("Please enter a non-negative number: "))

tolerance = 0.0001
if the_square == 0:
    approx_sqrt = 0.0
else:
    guess_low = 0.0
    if the_square >= 1:
        guess_high = the_square
    else:
        guess_high = 1.0

    while guess_high - guess_low > tolerance:
        midpoint = (guess_low + guess_high) / 2
        if midpoint ** 2 > the_square:
            guess_high = midpoint
        else:
            guess_low = midpoint

    approx_sqrt = (guess_low + guess_high) / 2

print("Approximate square root of", the_square, ":", approx_sqrt)
```
