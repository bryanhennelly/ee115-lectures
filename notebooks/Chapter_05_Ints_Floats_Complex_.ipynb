## 5.1 Integers (`int`)

Integers, often referred to as "ints," represent whole numbers—numbers without any fractional or decimal component. They can be **positive**, **negative**, or **zero**, and are fundamental for counting, indexing (don't worry what this means just yet!), and a variety of mathematical operations like **+** and **-** that can act on integers to produce a new integer.

### Defining Integers

To define an integer, simply write a number without a decimal point. Here are some examples:


```python
positive_integer = 123
negative_integer = -456
zero = 0
print(positive_integer, negative_integer, zero)
```

```python
a = 6
b = 4

print(a + b)   # Addition: 10
print(a - b)   # Subtraction: 2
print(a * b)   # Multiplication: 24
print(a // b)  # Integer Division (quotient): 1 (This rounds down the result to the closest integer)
print(a % b)   # Modulus (remainder): 2
print(a ** b)  # Exponentiation: 1296
```

### Binary Representation of Integers

Internally, integers are stored in binary (base-2). Positive and negative integers are encoded differently:

- **Positive integers** are stored in straightforward binary format.
- **Negative integers** use a method called **two's complement** encoding, where the most significant bit represents the sign of the number.

For example:
- `5` in binary is `101`
- `-5` in binary (using two's complement) is represented as `11111011` in 8-bit form.

Understanding how integers are stored can be useful in certain programming contexts, especially in optimization and working with low-level systems. #### C:

On many systems, an int might be 32 bits (4 bytes) which means it can hold a value from -2,147,483,648 to 2,147,483,647. A long long type typically provides at least 64 bits.

## 5.3 Floats (`float`)

Floats in Python represent real numbers that include a decimal point.

### Defining Floats:
To define a floating-point number in Python, simply use a decimal point or employ the scientific notation:

```python
simple_float = 20.5
small_float = -0.003
scientific_notation = 4.5e2  # Represents 4.5 * 10^2 = 450.0
print(simple_float, small_float, scientific_notation)
```

### Basic Operations with Floats:
Floating-point numbers support basic arithmetic operations just like integers.

```python
x = 7.2
y = 3.0

print(x + y)  # Addition: 10.2
print(x - y)  # Subtraction: 4.2
print(x * y)  # Multiplication: 21.6
print(x / y)  # Division: 2.4
```

### Precision Issues with Floats:
Computers use binary (base-2) to represent floating-point numbers, which can sometimes lead to unexpected results when trying to represent base-10 decimal numbers. This is a universal challenge seen in computing, not just in Python.

```python
print(0.1 + 0.2)         # Might not exactly print 0.3 due to rounding in binary representation
```

The reason for these unexpected results lies in how computers handle floating-point arithmetic. In the example above, neither 0.1 nor 0.2 (or the sum 0.3) can be represented exactly in binary. The stored values are the nearest representable binary fractions. When these near-miss values are summed, the result can be slightly off.

To ensure accurate computations, particularly in cases where precision is crucial:

- Consider using the decimal module in Python, which offers arbitrary precision.
- Round the numbers to a fixed number of decimal places before comparisons.
- Use libraries like NumPy that provide functions to compare floats within a desired tolerance.

Always be cautious and aware of potential pitfalls when working with floating-point numbers in any programming language.

#### IEEE 754 64-bit Floating-Point Representation

The IEEE 754 standard is used for representing floating-point numbers in many programming languages, including Python. In the 64-bit format (commonly referred to as **double precision**), the bits are divided as follows:

1. **Sign bit (1 bit)**: Determines whether the number is positive or negative. 
   - `0` means positive.
   - `1` means negative.

2. **Exponent (11 bits)**: Represents the exponent in **biased** form. The exponent is stored as an unsigned integer, with a bias of 1023. This allows both positive and negative exponents to be represented. The actual exponent `E` is computed as:
   \[
   E = \text{Exponent} - 1023
   \]

3. **Fraction (Mantissa) (52 bits)**: Represents the significant digits of the number, also known as the mantissa. The mantissa stores the digits of the number in binary, and there is an implicit leading bit (1), which is assumed to be in front of the binary fraction. This gives the mantissa a range between 1.0 and 2.0.

#### Structure Overview

Here's how the 64 bits are organized:

| **Sign** | **Exponent** (11 bits) | **Mantissa** (52 bits) |
|----------|------------------------|------------------------|
| 1 bit    | 11 bits                 | 52 bits                |

#### Example Breakdown

Let’s break down the representation of the floating-point number `-6.75`:

1. **Sign bit**: The number is negative, so the sign bit is `1`.

2. **Convert 6.75 to binary**:
   - 6 in binary is `110`.
   - 0.75 in binary is `0.11` (since 0.75 = 1/2 + 1/4).
   - So, `6.75` in binary is `110.11`.

3. **Normalize**: Normalize the binary number to scientific notation form:
   \[
   6.75 = 1.1011 \times 2^2
   \]
   The leading `1` is implicit in IEEE 754 format and is not stored explicitly in the mantissa.

4. **Exponent**: The exponent is `2`, but we store it as a biased exponent. The bias for 64-bit floating-point numbers is 1023:
   \[
   \text{Exponent} = 2 + 1023 = 1025 \quad \text{(which is `10000000001` in binary)}.
   \]

5. **Mantissa**: The fractional part `1011` (from the normalized `1.1011`) is stored in the mantissa (padding with 0s to fill 52 bits):
   \[
   \text{Mantissa} = 1011000000000000000000000000000000000000000000000000
   \]

#### Final Representation

Putting it all together:

- **Sign bit**: `1`
- **Exponent**: `10000000001`
- **Mantissa**: `1011000000000000000000000000000000000000000000000000`

Thus, the number `-6.75` is represented in 64-bit IEEE 754 format as:
\[
1\ 10000000001\ 1011000000000000000000000000000000000000000000000000
\]

#### Why This Matters

This system allows for a wide range of numbers, including very small and very large values, but it also introduces small precision errors for some numbers due to the limitations of representing decimal fractions in binary form. In summary, the IEEE 754 standard for floating-point arithmetic provides a balance between precision and range but comes with inherent limitations in representing certain decimal values exactly.


## 5.3 Complex Numbers (complex)

Complex numbers have a real part and an imaginary part.

Defining Complex Numbers:

```python
z1 = 2 + 3j
z2 = complex(3, -4)
print(z1)  # Outputs: (2+3j)
print(z2)  # Outputs: (3-4j)
```

Operations with Complex Numbers include the following:

```python
z1 = 2 + 3j
z2 = complex(3, -4)

print(z1 + z2)  # Addition: (5-1j)
print(z1 - z2)  # Subtraction: (-1+7j)
print(z1 * z2)  # Multiplication: (18+1j)
print(z1 / z2)  # Division: (-0.24+0.68j)
```

and the real and imaginary parts can be extracted as follows:

```python
z1 = 2 + 3j

print(z1.real)  # Outputs: 2.0
print(z1.imag)  # Outputs: 3.0
```

## 5.4 Type conversions

You can switch between numeric types using built-in functions called type conversion. 

```python
integer_value = 5
to_float = float(integer_value)
print(to_float)  # Outputs: 5.0

float_value = 8.7
to_int = int(float_value)
print(to_int)  # Outputs: 8 (Note: decimal part is truncated)
```

However, it's important to note that complex numbers cannot be directly converted to ints or floats. You must extract the real or imaginary parts first.

### Type Conversion with Strings in Python

In Python, data type conversion, also known as type casting or type coercion, allows for the conversion of a value from one data type to another. When dealing with strings, you'll often find situations where you need to convert between strings and other data types.

### String to Integer or Float:

If a string contains a number in a valid format, you can convert it to an `int` or `float` using the `int()` and `float()` functions, respectively.

```python
string_num = "123"
converted_int = int(string_num)
print(converted_int)  # Outputs: 123

string_float = "123.45"
converted_float = float(string_float)
print(converted_float)  # Outputs: 123.45
However, attempting to convert a non-numeric string will result in a ValueError.
```

### Integer or Float to String:

We've just learned about converting a string to a number. Converting numbers to strings is also straightforward, using the str() function.



```python
num = 123
converted_str = str(num)
print(converted_str)  # Outputs: "123"

floating_num = 123.45
converted_float_str = str(floating_num)
print(converted_float_str)  # Outputs: "123.45"
```

### Caution:
When converting between types, always make sure the conversion makes sense and is valid for the given value. For example, trying to convert a string with alphabetic characters to an integer will raise an error.

```python
invalid_conversion = int("hello")  # This will raise a ValueError
```

## 5.6 Advanced operators: Bitwise operators with integers

Bitwise operators act on the bit-level representations of integers. There are four of interest

- Bitwise AND (&)
- Bitwise OR (|)
- Bitwise XOR (^)
- Bitwise NOT (~)
- Left Shift (<<)
- Right Shift (>>)

Each of these operations allows for powerful manipulations of integers in Python. Here are some example to help understand the behavior of each to apply them correctly in mathematical and algorithmic contexts.

Bitwise AND (&)

```python
a = 12  # Binary: 1100
b = 10  # Binary: 1010
result = a & b
print(result)  # Outputs: 8 (Binary: 1000)
```

Bitwise OR (|)

```python
a = 12  # Binary: 1100
b = 10  # Binary: 1010
result = a | b
print(result)  # Outputs: 14 (Binary: 1110)
```

```python
Bitwise XOR (^)
```

```python
a = 12  # Binary: 1100
b = 10  # Binary: 1010
result = a ^ b
print(result)  # Outputs: 6 (Binary: 0110)
```

```python
Bitwise NOT (~)
```

```python
a = 12
result = ~a
print(result)  # Outputs: -13 (Negate and subtract 1)
```

Left Shift (<<)

```python
a = 5   # Binary: 0101
result = a << 1
print(result)  # Outputs: 10 (Binary: 1010)
```

Right Shift (>>)

```python
a = 20   # Binary: 10100
result = a >> 2
print(result)  # Outputs: 5 (Binary: 0101)
```
