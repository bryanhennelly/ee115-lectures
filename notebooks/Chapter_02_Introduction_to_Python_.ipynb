## 2.2 Python Basics: Keywords and Naming Rules

### What are Python Keywords?
In Python, keywords are special reserved words that have predefined meanings. They are words that already have a job in Python, so **we can't use them for other things like naming our variables or functions**. 

For example, words like `if`, `else`, and `return` tell Python to do specific things.

**Note**: Only three Python keywords have capital letters: `True`, `False`, and `None`. All other keywords are written in lowercase.

### Common Python Keywords You Should Know:

Here are some of the most commonly used Python keywords:

- **Boolean Values**: `True`, `False`, `None`
- **Control Flow**:
  - Conditional statements: `if`, `else`, `elif`
  - Loops: `for`, `while`
  - Exiting loops: `break`, `continue`
- **Functions and Classes**:
  - Defining functions: `def`, `return`
  - Defining classes: `class` (we wont use this until Semester 2 when we learn about object-oriented programming)
- **Logical Operators**: `and`, `or`, `not`, `is`, `in`
- **Error Handling**: `try`, `except`, `finally`

Don't worry about understanding these or remembering any of them now. As you learn Python, you'll beome familiar with all of these!

**Tip**: If you want to explore more keywords and their usage, you can check out this [comprehensive list of Python keywords](https://www.programiz.com/python-programming/keyword-list).


## 2.3 Understanding Python Identifiers

In Python, identifiers are essentially names we assign to entities like variables and other objects. Here's an example:

```python
language = 'Python'
```

In this code, `language` is an identifier that stores the value 'Python'.

Remember, Python has certain reserved words called "keywords" that have special meanings. These keywords **cannot** be used as identifiers. Here's what **not** to do:

```python
continue = 'Python'  # Incorrect!
```

This code is problematic because `continue` is a reserved keyword in Python. For more details, refer to [Python Variables](#).

---

### Guidelines for Naming Identifiers

- **Avoid Keywords**: Keywords in Python are off-limits for naming variables or other entities.
- **Case Matters**: Python treats `Hello` and `hello` as two completely different identifiers.
- **Start Correctly**: Identifiers can contain letters and numbers, but they must begin with a letter or an underscore (`_`). Starting with a number is not allowed.
- **Prefer Letters Over Underscores**: It is more conventional to start identifiers with a letter rather than an underscore.
- **No Spaces**: Identifiers cannot contain spaces.
- **Keep it Clean**: Special symbols like `!`, `@`, `#`, `$`, etc., should not be used in identifiers.
- **Use Meaningful Names**: Whenever possible, choose descriptive names like `name` if the variable is storing someone's name. This makes your code easier to understand.

---

### Examples of Good and Bad Identifiers

**Good Identifier Choices**:
- user_name
- total_score
- email_address
- product_list
- order_status
- item_count
- user_feedback
- max_limit
- min_value
- database_connection

**Poor Identifier Choices**:
- 123user (starts with a number)
- email@address (contains a special character)
- $price (starts with a special character)
- total#count (contains a special character)
- product list (contains a space)
- order&status (contains a special character)
- %feedback (starts with a special character)
- !limit (starts with a special character)
- min-value (contains a hyphen instead of an underscore)
- user&name (contains a special character)

---

### Remember This!

- Python is case-sensitive, so `apple` and `Apple` are considered different identifiers.
- Always aim for meaningful names when creating identifiers. For instance, `days_in_month` is much more intuitive than simply `d`.
- For identifiers with multiple words, connect them using underscores, like `first_name`.



## 2.4 Python Comments: Making Code Clearer

Comments in programming are like little notes you leave for yourself or other developers. They're there to explain parts of the code. The best part? Python doesn't act on them - they are not seen as code (instructins for the CPU). It's just for us humans. Here is a simple example:

````python
# Here, I'm setting up some numbers
a = 4
b = 7

# Let's see what they look like together
print('The numbers are:', a, ' and ', b)
````


From this, you can see comments start with a `#` and help describe what's happening.

### Different Kinds of Comments in Python

Python mainly has two styles of comments:

1. **Short Comments**: These are one-liners, and you use `#` to start them.

   ```python
   # This is a short comment
   fruit = 'apple'  # This comment is next to the code
   ```

2. **Long Comments**: Python doesn't have an official way for these, but we have tricks! You can either slap a `#` before each line or use three quotes (`'''` or `"""`).

   ```python
   # This is a way
   # to write longer comments
   # across many lines

   '''
   This is another method 
   for really long comments.
   It's kind of a hack, but it works!
   '''
   ```

![Figure 2.3:Two ways to comment](https://raw.githubusercontent.com/bryanhennelly/Programming-with-Python/main/comment.png)

### Why Bother With Comments?

1. **Understanding**: With good comments, anyone (including future you) can understand your code faster.
  
2. **Debugging Aid**: Got a line causing trouble? Turn it into a comment to stop it from running and causing errors. 

   ```python
   print('Hello World!')

   # print(This line is broken

   print('More code here.')
   ```

Remember, comments should be helpful. If you've got a tricky piece of code, use a comment to explain why you did it, not just what it does. Let your code show the 'what', and your comments show the 'why'.

## 2.5 Understanding Python Basics: Variables, Constants, and Values

### Variables in Python

In programming, think of a variable as a box where you can store data, such as numbers or words. For example:

```python
num = 20
```

Here, `num` is the box, and `20` is the data inside. When you want to see what's in the box, you can simply use its name, like `print(num)`. This process of assigning a value to a variable is called **Assignment**.

![Figure 2.5:Making a variable](https://raw.githubusercontent.com/bryanhennelly/Programming-with-Python/main/variable.png)

Note that we call these variable becuase I can reuse the name whenever I like....

```python
num = 20
print(num)
num = 50
print(num)
``

#### Setting a Variable's Value

In Python, the `=` symbol is used for assignment, i.e. to place data inside the box:

```python
name = 'Bryan'
print(name)
```

What’s great about Python is that you don’t need to specify what type of data you’re storing. Python figures it out for you! In other languages, such as Java or C++, you have to declare the data type.

#### Changing the Value of a Variable

```python
name = 'Bryan'
print(name)

name = 'Bill'
print(name)
```

Initially, the variable `name` held the string `'Bryan'`. We then changed its value to `'Bill'`.

#### Quick Tricks with Variables

**Simultaneous assignment**: You can assign values to multiple variables at once:

```python
x, y, z = 3, 4.5, 'Hi'
print(x)
print(y)
print(z)
```

**Collective assignment**: Or, if you want two variables to have the same value:

```python
box1 = box2 = 'sameStuff.com'
print(box1)
print(box2)
```

### Naming Rules for Python Variables

1. Use only letters, numbers, or underscores.
2. Start with a letter or underscore.
3. Make it meaningful: `username` is better than `u`.
4. Use lowercase (`my_var`), UPPERCASE (`MY_VAR`), CamelCase (`MyVar`), or even MixN_Match.
5. Python is case-sensitive (`Age` and `age` are different variables).
6. Avoid using reserved words (like `if` or `while`).

### Constants: The Locked Boxes

Constants are variables whose values shouldn’t change. Think of them as locked boxes. While Python allows you to change their values, by convention, you should treat them as unchangeable. Constants are typically written in UPPERCASE.

In Python, constants are often declared in a separate module (a file containing variables, functions, etc., that can be imported). Let’s see how to declare constants in a separate file and use them in the main code.

Create a `constants.py`:

```python
# declare constants
PI = 3.14
GRAVITY = 9.8
```

Now, you can import this file into another Python script:

```python
# import the constants file
import constants

print(constants.PI)       # prints 3.14
print(constants.GRAVITY)   # prints 9.8
```

### Python Values: Different Data Types

Values can be simple:

- Words or sentences: `'Hello!'` (this type of data is called a `string`)
- Whole numbers: `5, -15` (this is an `int`)
- Decimals: `3.14` (this is a `float`)

We’ll cover more complex data types later. For now, let’s stick to `string`, `int`, and `float`.

### Static vs Dynamic Type Declaration

In some programming languages, you need to specify the type of data a variable will hold when you declare it. Let’s compare Java and Python code for this:

**Static (Java):** You must declare the type of data.

```java
int myNumber = 10;
String myString = "Hello, Java!";
```

**Dynamic (Python):** Python determines the type automatically.

```python
my_number = 10
my_string = "Hello, Python!"
```

In Java, you specify `int` for numbers and `String` for text. In Python, you simply assign a value and Python takes care of the rest.

### Numeric Data Types in Python

Python supports several numeric data types:

- **`int`**: Represents integers, which can be of unlimited length.
- **`float`**: Represents floating-point numbers (decimals).
- **`complex`**: Represents complex numbers, which include real and imaginary parts.

You can use Python’s `type()` function to check the data type of any variable:

```python
num_integer = 8
print(num_integer, 'is of type', type(num_integer))  # Output: 8 is of type <class 'int'>

num_float = 2.5
print(num_float, 'is of type', type(num_float))  # Output: 2.5 is of type <class 'float'>

num_complex = 3 + 4j
print(num_complex, 'is of type', type(num_complex))  # Output: (3+4j) is of type <class 'complex'>
```
