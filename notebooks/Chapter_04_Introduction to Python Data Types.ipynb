## 4.1 Variable Assignment and Type

### A simple start..

Let's start with a simple one:

```python
x = 5
```
Here, the number 5 **is** an integer, **and** therefore, ```x``` becomes an integer variable.

![Figure 4.2:A simple assignment](https://raw.githubusercontent.com/bryanhennelly/Programming-with-Python/main/x_equals_5.png)

We can see here in this image that in Python when we write the smple line of code ```x = 5``` we have created an integer object and placed the value ```5``` inside it. The name ```x``` points to this object and whenever we use the name ```x``` on the **right-hand side** of an ```=``` sign we are effecively fetching the value stored in the memory location that ```x``` points to. When we make an integer object like this Python also makes a little note beside the nox saying this is an ```int``` so that it will understand what ```+``` and ```-``` and other operators will mean for this type of object. 
### Numeric Data Type in Python

Python provides a range of other numeric data types too:

- `int`: Represents signed integers of unlimited length.
- `float`: Represents floating-point real values.
- `complex`: Represents complex numbers.

To determine a variable's class, Python offers the `type()` function. Here are some example. Click to run the code below and see the output. 


````python
num_integer = 8
print(num_integer, 'is of type', type(num_integer))

num_float = 2.5
print(num_float, 'is of type', type(num_float))

num_complex = 3+4j
print(num_complex, 'is of type', type(num_complex))

my_string = 'Bryan'
print(my_string, 'is of type', type(my_string))
````

Note that the + and - and other operators might mean something different for different types of object. Also the number of bytes of memory put asside for the box, and the manner in which the ```1```'s and ```0```s encode this data will also be different for the different types.

### Using `type()` to tell what type of object something is

In our example, four variables are declared: `num_integer`, `num_float`, `num_complex`, and `my_string` with respective values 5, 5.0, 1+2j, and 'Bryan'. Using the `type()` function, we can identify the class or data type of each variable.

```python
num_integer = 5
```
`num_integer` holds the value `5`, which is an integer. Thus, when we run `type(num_integer)`, it yields `<class 'int'>`.

```python
num_float = 5.0
```
For `num_float`, the value is `5.0`, a floating-point number. Consequently, running `type(num_float)` gives `<class 'float'>`.

```python
num_complex = 1 + 2j
```
For `num_complex` contains `1+2j`, a complex number. Running `type(num_complex)` will display `<class 'complex'>`.

```python
my_string = 'Bryan'
```

Lastly `my_string` contains `'Bryan'`, a string. Running `type(my_string)` will display `<class 'string'>`.


## 4.2 Strings

### Strings

Strings represent a sequence of characters. In essence, they are words or sentences but they can be just one character or even no characters at all! Strigs are one of the foundational data types for many programming tasks. From representing words and sentences to holding data for processing, strings are versatile and essential. This guide will introduce you to the basics of working with strings in Python.

#### Single Quote

Strings can be defined using single quotes. For instance:

```python
'Quote me on this'
```

All spaces and tabs inside the quotes are retained.

#### Double Quotes

Double quotes function similarly to single quotes:

```python
"What's your name?"
```

#### Triple Quotes

For strings spanning multiple lines, triple quotes (`'''` or `"""`) are utilized. Both single and double quotes can be incorporated within triple quotes:

```python
'''
This is a multi-line string. This is the first line.
This is the second line.
"What's your name?," I asked.
He said "Bond, James Bond."
'''
```

````python
greeting = "Hello, World!"
print(greeting)
````

This code initializes a variable named `greeting` with the string "Hello, World!" and then prints it. Displaying information to the console is a fundamental task. In Python, the `print()` function enables us to output data to the terminal.

Another interesting thing about string objects is that we can access individual characters using square brackets ```[]``` as follows. Lets take this simple piece of code.

```python
language = "Python"
print(language)
```

### Unicode Characters

Python supports Unicode, a standard that allows for the representation of characters from virtually all written languages.

````python
# Greek alphabet example
alpha = "α"
print(alpha)

# Heart symbol example
heart = "\u2665"
print(heart)
````

This demonstrates that strings can hold a wide variety of characters, not just those in the English alphabet.

### Special Characters

Strings in Python can also contain special characters which are used to represent certain concepts:

- `\n` : Creates a new line
- `\t` : Inserts a tab

````python
print("First Line\nSecond Line")
print("Column1\tColumn2")
````

These characters allow for formatting within strings.

### Escaping Characters

Sometimes, we need strings to include characters that have a specific function in Python, like the double-quote in a string. To use these characters literally, we "escape" them with a backslash (`\`).

````python
sentence = "She said, \"Hello, World!\""
print(sentence)
````

The result is a printed string that includes the double quotes. 

### String Concatenation

Combining strings is a frequent operation. In Python, strings are concatenated using the `+` operator.


````python
str1 = "Hello"
str2 = "World"
combined = str1 + ", " + str2 + "!"
print(combined)
````

This will combine and print the strings as "Hello, World!". 

### The `format` method

When we want to combine literal strings and variables in a print statement we can do so with comma separation as follows:

````python
age = 44
name = 'Bryan'
print(name, ' was ', age, ' years old when he wrote this book')
print('Why is ', name, ' playing with that python?')
````

There is a better way to do this called the `format()` method, which can be used as follows:

````python
age = 44
name = 'Bryan'
print('{} was {} years old when he wrote this book'.format(name, age))
print('Why is {} playing with that python?'.format(name))
````

This method replaces each placeholder `{}` with the corresponding value. It's a more readable and less error-prone method than concatenating strings using `+` or using the comma separation in the print statement.

There are several ways to use the `format` method; we will learn about these a little later in the course.

There's even a better option...From Python 3.6 onwards, "f-strings" offer a concise way to embed expressions inside string literals:


````python
age = 44
name = 'Bryan'
print(f'{name} was {age} years old when he wrote this book')
````

## 4.3 This is the `end` of this chapter

The `end` parameter specifies what should be printed at the end of the given string(s). By default, this value is \n, which is a newline character. This means that by default, the print function adds a newline after printing the specified strings. You can override this by providing a different value for the end parameter.

Here's a brief example:

````python
print("Hello", end=", ")
print("World!")
````
