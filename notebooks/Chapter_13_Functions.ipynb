## 13.2 Pass a value into a function

### Example: Print multiples of a chosen number

The **parameter** `number` is a local name that receives the value supplied in a call. In `print_multiples(5)`, the **argument** is 5, so `number` is 5 during that call. A later call supplies a new value. We write the loop only once.

````python
def print_multiples(number):
    for multiplier in range(1, 4):
        print(number * multiplier)

print_multiples(5)
print_multiples(7)
````

`number` receives the argument for the whole call. `multiplier` is the loop variable: it takes 1, 2 and 3 in turn inside that call. Parameter names are our choice; callers do not need variables with matching names.

## 13.4 Return a result instead of printing it

### Example: Add two numbers and use the result

`return` sends a value back to the caller and ends the function call. Here `a` receives 10 and `b` receives 20; the returned value 30 is assigned to `answer`. We can then use that value in another calculation.

````python
def add(a, b):
    return a + b

answer = add(10, 20)
print("Answer:", answer)
print("Twice the answer:", answer * 2)
print("Another call:", add(4, 7))
````

### Example: See the difference between print and return

`print()` displays something. It does not send that displayed number back as the function result. A function that reaches its end without returning a value returns `None`, meaning no result value was supplied.

````python
def print_sum(a, b):
    print(a + b)

result = print_sum(2, 3)
print("Returned value:", result)
````

Printing functions are useful for displaying information; returning functions are useful when later code needs the result. A function may accept parameters whether or not it returns a useful value. A plain `return` also ends the call and returns `None`.

## 13.5 Put a familiar loop inside a function

### Example: Return the sum from 1 to n

This is the same accumulation loop used earlier, now reusable for any non-negative integer `n`. Initialise `total` for each call. Place `return` **after** the loop so all additions finish first.

````python
def sum_to_n(n):
    total = 0
    for number in range(1, n + 1):
        total += number
    return total

print(sum_to_n(5))
print(sum_to_n(100))
print(sum_to_n(0))
````

## 13.6 Return True or False

### Example: Check whether a number is prime

A prime is an integer above 1 with no divisors other than 1 and itself. Return `False` as soon as we know the number is not prime. Return `True` only after all possible divisors have been checked. **`return` leaves the whole function**, even from inside a loop.

````python
def is_prime(number):
    if number <= 1:
        return False
    for divisor in range(2, number):
        if number % divisor == 0:
            return False
    return True

print(is_prime(7))
print(is_prime(9))
print(is_prime(1))
print(is_prime(2))
````

### Example: Use the returned Boolean in a decision

The caller decides what to do with the result. We reuse `is_prime` from the preceding cell; run that definition first.

````python
number = 13
if is_prime(number):
    print(number, "is prime")
else:
    print(number, "is not prime")
````

## 13.7 Pass a list into a function

### Example: Return the minimum element of a list

The parameter `numbers` receives the supplied list. Start with its first value and keep the smallest found so far. This function requires a **non-empty** list.

````python
def find_min(numbers):
    minimum = numbers[0]
    for number in numbers:
        if number < minimum:
            minimum = number
    return minimum

readings = [3, 1, 4, 9, 2]
print(find_min(readings))
print(find_min([-5, -2, -8]))
````

## 13.8 Call one function from another

### Example: Use a total function to calculate an average

`average_list` calls `sum_list` and uses its result. This keeps the summing code in one place. Both definitions must have run before the call to `average_list`; the average requires a non-empty list.

````python
def sum_list(numbers):
    total = 0
    for number in numbers:
        total += number
    return total

def average_list(numbers):
    return sum_list(numbers) / len(numbers)

print(sum_list([10, 12, 14]))
print(average_list([10, 12, 14]))
print(sum_list([]))
````

### Example: Use smaller functions to calculate standard deviation

Standard deviation measures spread around the mean. For the complete list, we use the **population** formula: square root of the average squared distance from the mean. This reuses `average_list` twice.

````python
def std_deviation(numbers):
    mean = average_list(numbers)
    squared_differences = []
    for number in numbers:
        squared_differences.append((number - mean) ** 2)
    return average_list(squared_differences) ** 0.5

print(std_deviation([2, 4]))
print(std_deviation([5, 5, 5]))
````

## 13.9 Return two related results

### Example: Return both the minimum and maximum

A comma lets us return two values together. The assignment `low, high = ...` unpacks them in the same order. Technically the returned pair is a tuple; here we simply give its two values separate names. The input list must be non-empty.

````python
def find_min_and_max(numbers):
    minimum = numbers[0]
    maximum = numbers[0]
    for number in numbers:
        if number < minimum:
            minimum = number
        if number > maximum:
            maximum = number
    return minimum, maximum

low, high = find_min_and_max([3, 1, 4, 9, 2])
print("Minimum:", low)
print("Maximum:", high)
````

## 13.11 Passing values and references: what can a function change?

**When we pass a list into a function, we pass a reference to that list—not a new list. There is one list, with two names in two places: the caller’s name and the function’s parameter name.** Changing its elements through either name changes the same list.

You will encounter the terms **pass by value** (receiving a copy of a value) and **pass by reference** (accessing the caller’s variable through a reference). Python’s precise rule is **passing object references**, also called *object sharing*: the parameter receives a reference to the supplied object. Reassigning that parameter does not reassign the caller’s variable.

For a number, assigning a new value inside the function leaves the caller’s variable unchanged. For a list, changing an element or appending changes the shared list. The following examples show the difference.

### Example: Reassign an integer parameter

Python binds the parameter name to the object supplied by the caller; it does not automatically make a copy. `value = value + 10` binds the **local name** to the result of the addition. It does not reassign the caller’s `number`.

````python
def modify_value(value):
    value = value + 10
    print("Inside:", value)

number = 5
modify_value(number)
print("Outside:", number)
````

### Example: Change the list passed into a function

In `append_reading(readings, 4)`, the caller’s name `readings` and the parameter `numbers` refer to **one shared list**. No copy is made. The function uses its local name `numbers` to append to that list; afterwards, the caller sees the extra element through `readings`.

````python
def append_reading(numbers, value):
    numbers.append(value)

readings = [1, 2, 3]
append_reading(readings, 4)
print(readings)
````

### Example: Reassign a list parameter instead

Assigning a different list to the parameter changes only the local name. This is different from changing an element of the original list.

````python
def replace_list(numbers):
    numbers = [0, 0]
    print("Inside:", numbers)

readings = [1, 2, 3]
replace_list(readings)
print("Outside:", readings)
````

**Remember: one list, two names.** `numbers.append(...)` and `numbers[index] = ...` change the shared list. But `numbers = [...]` merely makes the local parameter refer to a different list; the caller still refers to the original.

Integer objects cannot be changed in place. An assignment such as `value = value + 10` gives the local name a new value; return that value and assign it in the caller if you want to update the caller’s variable.

## 13.12 Return a new list and preserve the original

### Example: Make a sorted working copy

Copy the list before changing it, then return the copy. For this flat list of numbers, `.copy()` is sufficient. Notice which list we sort.

````python
def sorted_copy(numbers):
    working = numbers.copy()
    working.sort()
    return working

original = [3, 1, 2]
ordered = sorted_copy(original)
print("Original:", original)
print("Ordered:", ordered)
````

## 13.13 Return a complete sequence

### Example: Return an arithmetic sequence

Move sequence-building into a function. `start` is the first term, `step` is the amount added each time, and `count` is the number of terms. Assume `count` is a non-negative integer.

````python
def arithmetic_sequence(start, step, count):
    terms = []
    term = start
    for position in range(count):
        terms.append(term)
        term += step
    return terms

print(arithmetic_sequence(2, 3, 5))
print(arithmetic_sequence(2, 3, 0))
````

## 13.14 Recursion: a function calls itself

### Example: Calculate factorial using a loop

For a non-negative integer, `n!` is the product of 1 through `n`; `0!` is defined as 1. Start the product at 1. This is an **iterative** solution: repetition comes from a loop.

````python
def factorial_loop(n):
    product = 1
    for number in range(1, n + 1):
        product *= number
    return product

print(factorial_loop(4))
print(factorial_loop(0))
````

### Example: Calculate the same factorial using recursion

A **recursive** function calls itself on a smaller problem. The **base case**, `n == 0`, returns directly and stops further calls. Otherwise, use `n! = n × (n−1)!`. Assume a non-negative integer input.

````python
def factorial(n):
    if n == 0:
        return 1
    return n * factorial(n - 1)

print(factorial(4))
print(factorial(0))
````

For `factorial(3)`, the calls go down to the base case, then their results return:

| Call | Work needed | Result |
|---|---|---|
| `factorial(3)` | `3 * factorial(2)` | 6 |
| `factorial(2)` | `2 * factorial(1)` | 2 |
| `factorial(1)` | `1 * factorial(0)` | 1 |
| `factorial(0)` | Base case | 1 |

Both versions achieve the same result. Recursion is another way to express repetition, not a requirement to replace loops. Every recursive step must move towards a base case; use small inputs here.

## 13.15 Recursive Fibonacci numbers

### Example: Return one Fibonacci number

Number the sequence from zero: `F(0)=0`, `F(1)=1`, and each later value is the sum of the previous two. The argument is the **index of one term**, not how many terms to return. Assume a non-negative integer.

````python
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)

print(fibonacci(0))
print(fibonacci(1))
print(fibonacci(6))
````

For example, `fibonacci(4)` needs `fibonacci(3) + fibonacci(2)`. This simple version recalculates many terms, so keep indices small. Building the sequence with a loop is usually more efficient when we need many terms.
