## 8.1 Introduction: Why the `while` loop?

In programming, one frequently encounters scenarios that require repetition. Just as you wouldn't manually write the same note a hundred times by hand, you wouldn't want to write repetitive lines of code. The `while` loop in Python offers an elegant mechanism to handle such repetitiveness. As we proceed in this chapter, we'll juxtapose the `while` loop with something you're already familiar with: the `if` statement.

### Understanding the Concept of Loops

Imagine a scenario where you need to relay a message 100 times. Rather than tediously typing out 100 print statements, loops come to the rescue. They allow you to condense those repetitive tasks into a succinct and efficient block of code.

### It's a little bit like using `if`

The `while` loop is an important part of every computer programming language (not just python!!) and it operates on a simple principle: keep running a block of code as long as a given condition holds `True`.

The structure of the `while` loop is:

```python
while condition:
    # code to be executed
```

Let's compare this to the `if` statement that you have already met.

```python
if condition:
    # code to be executed
```

They both look the same. In many ways they are the same:
- **`condition`**: Just like in the `if` statement, this condition is the heart of the `while` loop. Remember for the `if` statement the condition is **tested as True or False**, and if this was `True` the body of the `if` would be executed and if it was `False` it would be ignored. For the `while` loop we see the same thing happen: if the condition is `True` the body of the `while` will be executed and if it is `False` it would be ignored. The difference is that the body will be executed once, then twice, then again and again and again as long as the condition is `True`. The condition is checked before every loop iteration. In the basic loops below, an update eventually makes the condition `False`; the loop then stops and execution moves to the next line after the loop. Later we will also meet `break`, which can stop a loop early.
- **Loop body**: Similar to the body of the `if` statement, this contains the set of instructions to be executed with each loop iteration.


Tips:
- <span style="color:red">Drawing a parallel with the `if` statement: the `if` statement checks a condition and runs its body once if the condition is `True`. The `while` loop, on the other hand, keeps running its body repeatedly as long as its condition remains `True`.</span>

- <span style="color:red">It's paramount to design the loop's condition such that it will turn `False` at some juncture. Failing to do so can result in an "infinite loop", making the program run ceaselessly.</span>

Let's start with a simple example:

````python
counter = 0

while counter < 5:
    print(f"The value of counter is: {counter}")
    counter += 1
````

When executed, this code segment will output the value of `counter` five times. With each iteration of the loop, the value of `counter` increments by 1, due to the `counter += 1` statement.

The heart of the loop, the condition (`counter < 5`), checks if the current value of `counter` is less than 5 before each iteration. It's vital to recognize that this condition changes with every cycle of the loop because of the increment operation. 

```
IMPORTANT NOTE: For these basic loops, the update must eventually make the condition False so that the loop stops. Later we will see how break can also stop a loop. In our example, once `counter` reaches the value of 5, the condition becomes `False`, and the loop ceases its execution.
```

## 8.3 Avoid Infinite Loops!!!

When working with loops in programming, one of the critical aspects to be wary of is inadvertently creating infinite loops. These loops continue indefinitely because their stopping condition is never met.

### Understanding Infinite Loops

An infinite loop arises when the condition governing the loop never evaluates to `False`. This can be a deliberate design choice for scenarios where a continuous operation is needed, like in some server processes where we are always on, always waiting for someone to connect. However, more often than not, infinite loops result from oversight and can lead to various issues, such as unresponsive software, escalated resource usage, and potential system crashes. Consider this straightforward yet illustrative example:

```python
# This demonstrates an intentional infinite loop.
while True:
    print("This message will keep displaying indefinitely!")
```

In the code above, the loop's condition (`True`) remains constant, meaning the loop has no reason to halt.

### A Practical Mistake: The Forgotten Increment

Let's look again ot our first while loop (the count to 5) and see how a common error can lead to infinite loops:


**Read this faulty example; do not run it as written.** The counter never changes.

```python
counter = 0
while counter < 5:
    print(f"The current value of counter is: {counter}")
```

In this scenario, the omission of `counter += 1` means the value of `counter` remains at `0`. Thus, the condition `counter < 5` will always be `True`, resulting in the print statement executing indefinitely.

### How To Avoid Accidental Infinite Loops

To safeguard against unintentional infinite loops:

1. **Mindful Loop Design**: Be vigilant!! As you construct a loop, ensure you have a clear understanding of its starting condition, its ending condition, and how the loop progresses from start to end.
   
2. **Small-Scale Testing**: Before unleashing a loop meant for extensive iterations, give it a trial run with limited cycles. If you aim to loop a thousand times, first test with just ten to gauge its performance.

## 8.4 Using `while` Loops with User Input

The `while` loop can be very useful in scenarios where we need to continually prompt the user until a valid input is obtained. It provides an intuitive mechanism to validate and re-request data from users.

Consider the following example:

````python
name = ""

while name == "":
    name = input("Please enter your name: ")

print(f"Hello, {name}!")
````

Here, the loop keeps prompting the user for their name as long as they provide an empty input (i.e., they just press the Enter key without typing anything). Only when a non-empty string is entered does the loop exit, and the greeting message gets printed.

## 8.5 Incorporating Logical Operators in `while` Loops

You're already familiar with logical operators like `and`, `or`, and `not`. These operators can be seamlessly integrated into `while` loops to create more intricate conditions.

Let's delve into an example where we solicit the user for a number within a specific range:

````python
number = 0

while number < 1 or number > 10:
    number = int(input("Enter a number between 1 and 10: "))

print(f"You entered: {number}")
````

In this illustration, the loop continues to prompt the user until a number between 1 (inclusive) and 10 (inclusive) is provided. The logical `or` operator ensures that both conditions—being less than 1 or greater than 10—cause the loop to continue.

## 8.6 A Practical Example of `while`: Adding All Numbers Between 1 and 100

To truly understand the power of the `while` loop, let's walk through a classic mathematical problem: adding all the numbers between 1 and N. One approach to calculate the sum of all numbers from 1 to 100 is by utilizing a `while` loop as follows:

````python
#initalise our variables
total = 0
number = 1

while number <= 100:
    total += number #total gets bigger with every loop
    number += 1 #number increases by 1 with each loop

print(f"The sum of numbers from 1 to 100 using a loop is: {total}")
````

Give the code a try - now try changing the 100 value to be even bigger - try 10000!

The code operates as follows:

- We begin with a `total` value of 0, which accumulates the sum.
- The `number` variable, starting at 1, increments during each iteration.
- The loop executes until `number` surpasses 100.
- In each loop cycle, the current value of `number` is added to `total`.
- Once the loop completes its run, the final sum resides in the `total` variable.

### Warning: Running a loop thousands of time might not be the best way

Clearly loops can colve this problem; however there exists a more streamlined solution. This efficient method comes from a tale of the young Carl Friedrich Gauss. Legend has it that as a schoolboy, Gauss was tasked with adding all the numbers from 1 to 100 as a form of punishment by his teacher, expecting it to take a while. However, young Gauss quickly recognized a pattern and immediately wrote down the answer, stunning his teacher. Gauss observed that if you pair the first and last number, then the second and the second-to-last number, and so on, each pair sums to 101. With 50 pairs, the total is  50 times 101 = 5050.

Using Gauss's insight we can get the answer with just one line of code:

````python
# A simpler way - less computation
n = 100
sum_gauss = n * (n + 1) // 2
print(f"Using Gauss's insight, the sum is: {sum_gauss}")
````

In just a single computation, Gauss's method bypasses the need to loop 100 or even 10000 times. This story underlines not only the genius of Gauss but also the beauty of efficient problem solving in mathematics and programming. In summary, while loops are instrumental tools in computing, it's beneficial to remain open to alternative, often more elegant, solutions.

## 8.7 Nested `while` Loops

In programming, it's often necessary to use loops inside other loops. This is known as nesting. Just like how you can nest `if` statements within other `if` statements, you can nest `while` loops within other `while` loops. When you nest loops, each iteration of the outer loop triggers the inner loop to complete all its iterations. This behavior can be particularly useful when working with multi-dimensional data or when performing repetitive tasks in a multi-tiered manner.

Let's explore this concept through a hands-on example:

````python
#Run the code to see what happens - play around with the values
x = 0

while x < 3:
    y = 0
    while y < 3:
        print(f"x: {x}, y: {y}")
        y += 1
    x += 1
````

In this example:

1. The outer loop manages the `x` variable, and for each iteration of `x`, the inner loop runs entirely for the `y` variable.
2. On the first iteration with `x` being 0, the inner loop will run three times with `y` taking values 0, 1, and 2.
3. Once the inner loop completes its cycle for all values of `y`, we move to the next iteration of the outer loop, incrementing the value of `x`.
4. This process continues until both `x` and `y` have taken all values less than 3.

As a result, the script prints combinations of `x` and `y` values for the range we've specified (in this case, both less than 3). This yields a total of 9 outputs, as both `x` and `y` can have 3 distinct values.

## 8.8 Utilizing `break` and `continue` in `while` Loops

In programming, especially when working with loops, there are times when we might want to alter the natural flow of a loop, either by skipping an iteration or by exiting the loop entirely before its condition becomes `False`. This is where the `break` and `continue` statements come in handy.

### `break` Statement

The `break` command acts as an immediate loop terminator. When executed, it halts the loop, regardless of whether the loop's condition still holds `True`.

### `continue` Statement

On the other hand, the `continue` command is subtler. Instead of terminating the loop, it bypasses the current iteration, moving directly to the next cycle.

Let's dive into an illustrative example:

````python
#Example of break and continue
counter = 0

while counter < 10:
    counter += 1
    if counter == 5:
        continue
    elif counter == 8:
        break
    print(counter)
````

The output for the above code would be: 
1
2
3
4
6
7

We can make several observations from the output:

- The number `5` is absent. When `counter` reaches 5, the `continue` statement gets triggered, skipping the print command for that iteration.
- The numbers after `7` are also missing. This is due to the `break` statement, which is executed when `counter` equals 8. It immediately halts the loop, thus, `8`, `9`, and `10` are never printed.

In essence, `break` and `continue` provide granular control over loop executions, allowing for more dynamic and responsive loop behaviors.
