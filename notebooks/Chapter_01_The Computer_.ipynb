## 1.1 What is a Computer?

Today's advanced computers, as illustrated in Figure 1.1, have a wide array of functionalities and capabilities. Whether you're diving into an immersive video game, drafting a detailed document, or editing a photograph, a computer stands ready to assist. At the heart of each of these tasks is a series of precise instructions that the computer must follow precisely.

These instructions guide the computer through various computational tasks. Such tasks may encompass reading from or writing to the computer's memory, conducting mathematical calculations, or executing logical operations. When these instructions are organized in a specific order to accomplish a particular objective, they collectively form what is referred to as a "program." In essence, every action a computer takes, from the simplest to the most complex, hinges on the execution of these programs.

As a programmer you write the instructions! You tell the computer what to do - step by step! One line of code at a time. And you have to write the code perfectly according to the rules that the computer expects it to be written. One letter wrong and CRASH!

## 1.2 The Von Neumann Architecture and Components

Highlighted in Figure 1.2 is the Von Neumann architecture, a model for the general purpose computers we interact with in our daily lives and the computer that you will learn to program in this course. This architecture offers insight into how a computer processes and understands the instructions you provide through your program. A central tenet of the Von Neumann design is its approach to memory: it houses not only the data but also the instructions, or programs. This concept is commonly referred to as the "stored-program computer."

![Figure 1.2: Von Neumann Architecture](https://raw.githubusercontent.com/bryanhennelly/Programming-with-Python/main/Figure1-2.png)

The Von Neumann architecture consists of three main components:

- __Memory__ - Stores instructions (programs) AND data.
- __A Central Processing Unit (CPU)__ Often termed the 'brain' of the computer, this is where all computations occur. The CPU follows the precise sequence of instructions it is given, which can involve:
    - Memory operations (reading or writing to/from memory)
    - Arithmetic and logic operations (additions, divisions, shifts...)
- __Input/Output (I/O) devices__ These bridge between the user and the computer. For instance, a keyboard allows data entry, so it is an input device, whereas a monitor, which displays results or graphical data, is an output device. Other input/output devices include hard drives (for storage), web cameras (for video input), memory cards or usb keys (for data transfer), and more, as depicted in Figure 1.3.

## 1.3 Memory

Data used by the computer are stored in the computer's memory. To access a specific piece of data, we use its address - the address of the data is a unique identifier for exactly where it is stored in the memory. Each memory location consists of 8 memory cells. Each cell stores one __bit__, and thus, each location stores 8 bits of information. Eight bits is equivalent to one __byte__.

### Bits and Bytes

A binary digit (bit) is the smallest unit of data in a computer, as depicted in Figure 1.4. A bit can store either a 0 or a 1, that is, it has 2 possible values. A byte is composed of 8 bits. Each of these 8 bits can store 2 possible values (0 or 1). How many possible values can a byte hold?

A byte can hold 28 = 256 possible values.

![Figure 1.4: Bits and Bytes:](https://raw.githubusercontent.com/bryanhennelly/Programming-with-Python/main/bits-and-bytes.png)

### Memory Organisation

A computer's memory size is set by the number of bits it uses for addresses. For instance, a computer with a 32-bit address system can have up to \(2^{32}\), which is 4GB of memory (calculated as \(4 \times 1024^3\)). Memory is grouped into units called **words**. A word can have 4, 8, or more bytes. These words move between the memory and the CPU using the system bus. There are lots of other jargon we use in compyter science to describe different bundles of bits, like 'nibble' for example. ### Memory Types in Computers

Computers use various types of memory, each designed for specific functions. Understanding the difference between these can help you grasp how computers process and store information.

#### Primary Memory
Primary memory, also known as main memory, is directly accessible by the Central Processing Unit (CPU). It acts like the computer's "working" or short-term memory, temporarily holding data and programs that the CPU is actively using. Here are the key types of primary memory:

- **Random Access Memory (RAM)**: This is the core component of primary memory. RAM allows data to be read and written in any order (hence "random access"). It's volatile, which means all data stored in RAM is lost when the computer is turned off. Think of RAM as a desk where a person works; you can easily access and change the contents on the desk, but once you clear it into the bin, everything is gone.

### Secondary Memory
Secondary memory is used for long-term data storage. It is not directly accessible by the CPU and is accessed via the computer's input/output channels. Types of secondary memory include:

- **Read Only Memory (ROM)**: ROM stores crucial data that does not change, like the firmware—the software permanently programmed into the computer. It's non-volatile, meaning it retains its contents even when the computer is turned off. ROM is analagous to a book in a library; you can read it as many times as you want, but its contents cannot be changed easily. 

- **Flash Memory**: Unlike ROM, flash memory can be erased and rewritten multiple times. It's a versatile type of non-volatile memory used in various devices like USB drives, memory cards, and solid-state drives (SSDs). Flash memory combines the speed of RAM with the permanence of ROM, making it ideal for storage solutions that require frequent updates.

Understanding these memory types is important in learning how computers operate efficiently, handling tasks from running applications to storing data long-term.

## 1.4 The Bus: Connecting Memory to the CPU

### What's a bus???

The main memory is not integrated into the CPU chip; instead, it resides on the computer's main board. Communication between the CPU and memory is facilitated by a **bus** system shown in Fig 1.6, which comprises several types of fast connecting wires:
- **Data Bus**: This bus carries the actual data (the ones as zeros represnted as eletrical pulses!) back and forth between the CPU and other hardware components.
- **Address Bus**: Functions like a mailing system, it carries the addresses of the memory locations where data is to be read from or written to. This bus is unidirectional, meaning it only sends information from the CPU to memory or other components, not the reverse.
- **Control Bus**: This is the traffic controller. It sends signals that manage the timing and execution of operations across the system, such as commands to read or write data. The control bus ensures that all parts of the computer operate in sync.

![Figure 1.6: The data bus carries bits between the CPU and the memory:](https://raw.githubusercontent.com/bryanhennelly/Programming-with-Python/main/data_bus.png)

### How fast is the bus???
Two features determine how much data can be moved accross the bus at speed:
- **Width**: The width of a bus is critical as it determines the volume of data that can be transferred concurrently in parallel. A wider bus can move larger blocks of data.
- **Clock Speed**: This is the speed at which data is transferred across the bus in MHz (that is millions of times per second). Higher clock speeds can significantly increase the rate at which data is processed.

## 1.5 The Central Processing Unit (CPU)

The Central Processing Unit (CPU), often called the brain of the computer, orchestrates all computer activities,  like a conductor leads an orchestra. It consists of several key components that handle specific functions.

### The Control Unit (CU)
The Control Unit (CU) directs CPU operations and manages the following three-step process, which is illustrated in Figure 1.7:
1. **FETCH**: Retrieves the next instruction from memory and stores it in the Instruction Register (IR).
2. **DECODE**: Interprets the instruction and determines the required action.
3. **EXECUTE**: Executes the instruction, managing data movement between registers, through the Arithmetic and Logic Unit (ALU), and back into memory for result storage.
![Figure 1.6: The CPU's Fetch, Decode, Execute process:](https://raw.githubusercontent.com/bryanhennelly/Programming-with-Python/main/Fetch_Decode_Exe.png)

### Clock System
The CPU includes a clock that synchronizes operations with a steady rhythm of electrical pulses. Clock speed, measured in hertz (Hz), dictates how many operations the CPU can perform per second. For example, a speed of 8GHz represents eight billion cycles per second, with higher speeds indicating a faster CPU.

### The Arithmetic and Logic Unit (ALU)
The ALU performs all arithmetic and logical operations within the CPU. It's essential for calculations and decision-making processes involved in computing.

### Registers: The CPU's Short-Term Memory
Registers are small, fast storage locations inside the CPU that hold data and instructions in use. Key registers include:
- **Program Counter (PC)**: Tracks the address of the current instruction.
- **Instruction Register (IR)**: Holds the instruction being decoded and executed.

### Cache Memory: Enhancing Performance
To avoid the slowdown of fetching data from main memory, frequently used data is stored in cache memory. This high-speed memory allows the CPU to quickly access data and instructions that are repeatedly needed, significantly improving performance.

## 1.6 Recap! Key Points


- **Von Neumann Architecture**: The foundation of modern computing, encompassing:
  - **CPU (Central Processing Unit)**: The central hub where all processing occurs.
  - **Memory**: Acts as storage for data and instructions.
  - **I/O Devices**: Facilitate data input into the system and output from it.


- **CPU Operations**: Executes a three-step cycle:
  - **Fetch**: Instructions are retrieved.
  - **Decode**: Instructions are interpreted.
  - **Execute**: Instructions are carried out.


- **Memory Hierarchy**:
  - **Primary Memory**: Directly accessible by the CPU, volatile.
  - **Secondary Memory**: Used for long-term storage, non-volatile.
  - **RAM**: Temporary storage that is fast but loses data when power is lost.
  - **ROM**: Permanent storage for critical data, not editable.
  - **Flash Memory**: Non-volatile and rewritable, used in USB drives and SSDs.


- **CPU Components**:
  - **Control Unit**: Manages the flow of data within the CPU.
  - **ALU (Arithmetic and Logic Unit)**: Performs all mathematical and logical operations.
  - **Registers**: Provide fast, accessible storage for immediate data needs.
  - **Cache**: Improves performance by storing frequently accessed data.


- **Buses**:
  - **Data Bus**: Transfers actual data between components.
  - **Address Bus**: Directs data to the correct location.
  - **Control Bus**: Manages the timing of operations and coordination among components.


- **Performance Metrics**:
  - **Bus Width**: Determines the amount of data transported at once.
  - **Clock Speed**: Affects how fast operations are processed.
