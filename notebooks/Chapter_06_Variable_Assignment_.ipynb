## 6.1 Introduction: Changing a Variable

In our previous discussions, we've approached Python variables as if their values remain the same throughout the execution of a program. In essence:

- After we initialize a variable with a value, it retains that value unless explicitly changed.
- This consistency makes our code predictable. Whenever we refer back to a variable, we can confidently expect it to hold the same value we initially assigned to it.

While many variables in our programs do remain constant throughout their lifecycle, there are scenarios where a variable's value needs to evolve and change from time to time. You will soon encounter this concept in the context of `for` and `while` loops. In this chapter, we'll explore two core mechanisms by which a variable's value can change:

1. **Variable Reassignment:** This is where a variable is explicitly given a new value.
2. **Object Mutation:** A subtler concept, where the content of an object is modified without reassigning the variable itself.

At the beginning of your programming journey you will only encounter **variable reassigment** and we will cover this today. In later chapters we will deal with **object mutation** so you can ignore this concept for now. Let's kick things off by diving deeper into variable reassignment.

## 6.2 Understanding Assignment and Reassignment in Python

In Python, a statement of the form `___ = ___` is known as an **assignment statement**. This statement:

- Uses a variable name on the left-hand side.
- Takes a literal value or an expression on the right-hand side.
- Assigns the value of the expression to the specified variable.

**Variable reassignment** is when we assign a new value to a variable that already has a value associated with it. This reassignment is most commonly performed using an assignment statement. Here's an example to illustrate this:


````python
x = 1
print(x)  # At this point, x refers to the value 1.
x = 5  # Here, the variable x is being reassigned.
print(x)  # Now, x refers to the value 5.
````

## 6.3 Variable reassignment never affects other variables. 

Let's analyze the following Python code:


````python
x = 1
y = x + 2
x = 7
print(x,y)
````

In the code above:

- `x` is initialized to the value 1.
- `y` is initialized to the value 3 (since `x` was 1 at the time of the assignment).
- `x` is then reassigned to the value 7.

An important question arises: **What is the value of `y` after `x` is reassigned?**

Python adheres to a fundamental principle:

> **Variable reassignment only impacts the immediate variable being reassigned and does not affect any other variables or values, even those initially defined using the variable undergoing reassignment.**

Given this principle, in our example above, the value of `y` remains 3, even after `x` is reassigned to 7.

This behavior might seem counterintuitive initially. However, it's actually a straightforward approach employed by Python. It ensures that developers can evaluate assignment statements sequentially, without concern that future reassignments might influence previous ones. When analyzing the code and encountering the statement `y = x + 2`, we can confidently determine the value of `y` based on the then-current value of `x`, irrespective of any future reassignments to `x`.

## 6.4 Augmented Assignment in Python

````python
sum_so_far = 10
x = 1
sum_so_far = sum_so_far + x # Reassign the value of sum_so_far by adding the current value of x and itself
print(sum_so_far)
````

The critical step happens in the third line of code. The expression `sum_so_far + x` computes the sum of 10 (the current value of `sum_so_far`) and 1 (the value of `x`), resulting in 11. This new value is then reassigned to the `sum_so_far` variable. At this juncture, the original value of `sum_so_far` (which was 10) is overridden and replaced by the new value (11).

Due to the frequent use of such reassignments in programming, Python has introduced a special shorthand syntax known as the "augmented assignment statement". Other programming languages also use similar shorthand notations. Augmented assignment statements can vary depending on the operation being performed. For example, the shorthand for adding a value to a variable is:

```python
<variable> += <expression>
```

So what does this code do? Lets break it down step-by-step:
    
1. The Python interpreter first checks if the variable exists and has a value. If not, it will give an error (called NameError). You can test this out in the Python console!
2. It then calculates the value of the expression on the right side.
3. Next, it adds the value of the variable to the calculated value from the previous step.
4. Finally, the variable is updated with this new sum.

For instance, instead of writing `sum_so_far = sum_so_far + x`, you can use the shorthand version with the same result: `sum_so_far += x`. Python has several more of these short cuts. Here is a list:

- `x += 3`  Equivalent to: x = x + 3
- `x -= 2`  Equivalent to: x = x - 2
- `x *= 4`  Multiply current value by 4: x = x * 4
- `x /= 2`  Divide current value by 2: x = x / 2
- `x %= 3`  Find remainder when divided by 3: x = x % 3
- `x **= 3` Raise to the power of 3: x = x ** 3
- `x //= 3` Integer division by 3: x = x // 3
- `x &= 3`  Bitwise AND with 3: x = x & 3
- `x |= 3`  Bitwise OR with 3: x = x | 3
- `x ^= 3`  Bitwise XOR with 3: x = x ^ 3
- `x <<= 2` Left shift by 2 bits: x = x << 2
- `x >>= 2` Right shift by 2 bits: x = x >> 2

## 6.5 Variable Swap

Suppose you have two variables, and you want to exchange their values, effectively making each variable hold the value of the other. This is a fundamental technique familiar to all programmers.

````python
a = 42
b = 13
print(f"Before swap: a = {a}, b = {b}")
a=b #Failed swap
print(f"After assignment: a = {a}, b = {b}")
````

You might think that just doing `a = b` would suffice. However, this approach has a flaw: it overwrites the original value of `a`, making it lost forever. Hence, the conventional method to solve this issue involves a temporary variable, typically named temp, to hold the value of a during the swapping process.

Here's the three-line solution using a temporary variable:

````python
a = 42
b = 13
temp = a #create a dummy variable to store a copy of a's value before we lose it
a = b
b = temp
print(f"After swap: a = {a}, b = {b}")
````

By following this sequence, `a` now has the value of `b`, and `b` gets the original value of `a`, successfully completing the swap.
