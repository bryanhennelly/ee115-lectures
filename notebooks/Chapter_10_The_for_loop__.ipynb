## 10.1 Repeating a task

You already know how to repeat code with `while`. A `for` loop gives a variable each value from a sequence, one at a time. With `range()`, Python takes care of moving to the next integer.

Here are two ways to print 1 to 5. Notice what disappears in the second version.

````python
count = 1
while count <= 5:
    print(count)
    count += 1

for count in range(1, 6):
    print(count)
````

## 10.2 What does `range()` give us?

`range(n)` supplies integers from **0 up to, but not including, n**. `range(n, m)` starts at n and stops before m. The default step is +1.

`print(range(4))` prints the description `range(0, 4)`, not the individual numbers. A loop lets us see those numbers:

````python
print(range(4))
for number in range(4):
    print(number)

print("Starting at 2:")
for number in range(2, 5):
    print(number)
````

## 10.4 When should we use `for` or `while`?

**Ask three questions:** we know the number of repetitions, so `for` fits.

**Ask until the user says yes:** the number of attempts is unknown, so `while` fits.

````python
for question in range(1, 4):
    answer = input("Type an answer: ")
    print("Answer", question, "received")
````

````python
answer = ""
while answer != "yes":
    answer = input("Ready? Type yes when ready: ")
print("Starting")
````

## 10.6 Searching with `break` and `continue`

### Is a number prime?

A prime is an integer greater than 1 with no positive divisors except 1 and itself. To test 17, try possible divisors from 2 to 16. A zero remainder means we found a divisor: mark the answer False and stop searching.

The Boolean `is_prime` records the conclusion of the loop.

````python
number = 17
is_prime = number >= 2

for divisor in range(2, number):
    if number % divisor == 0:
        is_prime = False
        break

print(number, is_prime)
````

With 17 the Boolean stays True. With 18 it becomes False as soon as divisor 2 is tested. For 0 or 1, it starts False; for 2 it starts True and the range is empty.

### Greatest common divisor (GCD)

A common divisor divides both numbers exactly. Search upwards and keep the latest match: the final match is the greatest. Here we use positive integers.

````python
a = 12
b = 18
limit = a
if b < limit:
    limit = b

gcd = 1
for divisor in range(1, limit + 1):
    if a % divisor == 0 and b % divisor == 0:
        gcd = divisor

print("GCD:", gcd)
````

### Least common multiple (LCM)

A common multiple is divisible by both numbers. Search upwards and stop at the **first** match. For positive a and b, their product is always a common multiple, so it gives us a finite search limit.

````python
a = 4
b = 6
start = a
if b > start:
    start = b

for candidate in range(start, a * b + 1):
    if candidate % a == 0 and candidate % b == 0:
        lcm = candidate
        break

print("LCM:", lcm)
````

### Skip an unwanted value with `continue`

`continue` skips the rest of the current iteration; the for loop still moves to its next value. This example adds 1 to 10 but leaves out multiples of 3.

````python
total = 0
for number in range(1, 11):
    if number % 3 == 0:
        continue
    total += number
print(total)
````

### Add a geometric series

Consider 1 + x + x² + x³ + … . For |x| < 1, adding more terms approaches 1 / (1 − x). For x = 0.5 the terms are 1, 0.5, 0.25, 0.125, … and the total approaches 2.

Eight terms means powers **0 to 7**, not 1 to 8.

````python
x = 0.5
terms = 8
total = 0
for power in range(terms):
    term = x ** power
    total += term
    print(power, term, total)
print("Value approached:", 1 / (1 - x))
````

### An alternating series

Changing + to − on alternate terms gives another useful pattern:

π / 4 = 1 − 1/3 + 1/5 − 1/7 + …

The denominators are odd numbers; multiplying sign by −1 switches it each time. This expansion converges slowly, so a short sum is only a rough approximation.

````python
terms = 10
sign = 1
total = 0
for index in range(terms):
    denominator = 2 * index + 1
    total += sign / denominator
    sign *= -1
print("Approximation to pi:", 4 * total)
````
