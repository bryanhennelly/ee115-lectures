# Chapter 11: Introduction to Lists

A **list** is a sequence of objects grouped under one name. Each object in the list is called an **element** (or item). The elements have a definite order: there is a first element, a second element, and so on.

We use lists to keep related objects together — for example, the names of tools, the lengths of cables, or information about an instrument. The elements can be numbers, strings, or other objects; they do not all have to be the same type.

## 11.1 Creating lists

Place the elements inside square brackets, separated by commas. Assign the list to a name that describes what it contains.


````python
tool_names = ["pliers", "screwdriver", "wire cutters"]
cable_lengths = [1.5, 2.0, 0.75]
instrument = ["voltmeter", 12, True]

print(tool_names)
print(cable_lengths)
print(instrument)
````

In `tool_names`, the first element is the string `"pliers"`. In `cable_lengths`, the elements are numbers. The list `instrument` groups three different types: a name, an equipment number, and an availability flag.

A list can also be empty: `[]`. We can add, remove, or change its elements, as we will see below.

## 11.2 Accessing items: the index

The first item has index **0**, the second has index **1**, and so on. Negative indices count from the end: `-1` is the last item. `len()` gives the number of items.

````python
lengths = [2.4, 1.8, 3.2, 0.9]

print("Second length:", lengths[1])
print("Last length:", lengths[-1])
print("Number of lengths:", len(lengths))
print("Last length again:", lengths[len(lengths) - 1])
````

### Stay within the list

For a list of length N, non-negative indices run from 0 to N − 1. An invalid index raises `IndexError`. Check the index before using it.

````python
equipment = ["meter", "power supply", "oscilloscope"]
index = 3

if 0 <= index < len(equipment):
    print(equipment[index])
else:
    print("Invalid index")
````

## 11.4 Adding and removing items

An empty list can grow as new data arrives.

````python
lengths = []
lengths.append(1.5)
lengths.append(2.0)

print(lengths)
print(len(lengths))
````

### Add at a position: `insert()`

`insert(index, item)` puts the new item before the item currently at that index. Later items move along.

### Remove by value: `remove()`

`remove()` deletes the first matching value. Use `in` to check that it exists; removing an absent value raises `ValueError`.

````python
tools = ["pliers", "meter", "pliers"]

if "pliers" in tools:
    tools.remove("pliers")

print(tools)
print("meter" in tools)
````

## 11.5 Sorting and reversing

`sort()` arranges comparable items in ascending order. `reverse()` reverses their current order. Both change the list itself; do not assign their return values back to the list.

````python
lengths = [2.5, 0.8, 1.6, 1.1]
lengths.sort()
print("Ascending:", lengths)

lengths.reverse()
print("Descending:", lengths)
````

### Two names or two lists?

Assignment binds a name to an object; it does not make a copy. In the first example, both names refer to the same list.

````python
original = [10, 20, 30]
alias = original
alias[0] = 99

print(original)
print(alias)
````

For a separate working list, use `.copy()`. Here its items are numbers, so changing an item in the copy leaves the original unchanged.

````python
original = [10, 20, 30]
working = original.copy()
working[0] = 99

print(original)
print(working)
````

## 11.6 Taking a slice

A slice extracts part of a list. As with `range()`, the stop is excluded. Omit the start to begin at the first item; omit the stop to continue to the end.

````python
stages = ["prepare", "connect", "measure", "check", "save", "finish"]

print(stages[1:4])
print(stages[:2])
print(stages[-3:])
````

### Choose a step

The full slice form is `[start:stop:step]`. A step of 2 selects alternate items; `[::-1]` makes a reversed list. These slices leave the original list unchanged.

````python
letters = ["A", "B", "C", "D", "E", "F"]

print(letters[::2])
print(letters[::-1])
print(letters)
````

## 11.7 A list of lists

A nested list can hold a table: each inner list is a row. In `table[row][column]`, both indices start at zero. One index selects the whole row.

````python
grid = [
    [1, 2, 3],
    [4, 5, 6]
]

print(grid[0])
print(grid[1][2])

grid[0][1] = 9
grid.append([7, 8, 9])
print(grid)
````

Appending to `grid` adds a row. Appending to `grid[0]` adds an item to the first row. For a rectangular measurement table, give every row the same number of readings.
