## 12.1 What is a loop variable?

**Goal: visit each number in a list and print twice its value.**

In `for number in numbers:`, `numbers` is the list and `number` is the **loop variable**. Python assigns the next list element to `number` before running the indented code.

On the first pass, `number` is 3; on the second it is 5; on the third it is 7. The indented code runs once for each element. After the last element, the loop ends.

````python
numbers = [3, 5, 7]

for number in numbers:
    print(number * 2)

print("Finished")
````

**Output**
```text
6
10
14
Finished
```

`number` is a name we choose; `value` would work too if we used it consistently. It holds the current **element**, not its index. Python sets it automatically: no initial assignment or `number += 1` is needed.

The unindented `print("Finished")` runs once, after the loop. The list itself remains `[3, 5, 7]`.

## 12.2 Add all the numbers in a list

### Example: Find the total cable length

Start `total` at zero **before** the loop. Each pass adds the current length to the running total. Print the result after the loop.

````python
lengths = [2, 3, 4]
total = 0

for length in lengths:
    total += length

print("Total length:", total)
````

**Output**
```text
Total length: 9
```

After the three passes, `total` is 2, then 5, then 9. Putting `total = 0` inside the loop would erase the previous additions.

## 12.3 Find the average of a list

### Example: Calculate the average temperature

For a non-empty list, average = total ÷ number of elements. First find the total using a loop, then divide by `len(temperatures)`.

````python
temperatures = [18, 20, 25]
total = 0

for temperature in temperatures:
    total += temperature

average = total / len(temperatures)
print("Average:", average)
````

**Output**
```text
Average: 21.0
```

## 12.4 Find the minimum and maximum values

### Example: Find the minimum: the smallest element

For this non-empty list, begin by treating the first element as the minimum. If a smaller value is found, replace the stored minimum. This also works when all values are negative.

````python
temperatures = [18, 15, 21, 12, 17]
minimum = temperatures[0]

for temperature in temperatures:
    if temperature < minimum:
        minimum = temperature

print("Minimum:", minimum)
````

**Output**
```text
Minimum: 12
```

`minimum` starts at 18, changes to 15, then to 12. It keeps the smallest value found so far. To find the **maximum** (the largest element), keep a value when it is **greater** than the current maximum.

## 12.5 Count values greater than a threshold

### Example: Count temperatures above 22 degrees

A threshold is a limit used in a comparison. Here we count values **strictly greater than** 22, so a reading equal to 22 is not counted.

````python
temperatures = [18, 25, 22, 27, 20]
threshold = 22
count = 0

for temperature in temperatures:
    if temperature > threshold:
        count += 1

print("Above threshold:", count)
````

**Output**
```text
Above threshold: 2
```

## 12.8 Update the elements stored in a list

### Example: See why changing the loop variable is not enough

Here `number` refers to each number in turn. Assigning a new number to that name does **not** replace the element stored in `numbers`.

````python
numbers = [3, 5, 2]

for number in numbers:
    number = number * 2

print(numbers)
````

**Output**
```text
[3, 5, 2]
```

### Example: Double the stored elements using their indices

Use `range(len(numbers))` to get indices 0, 1 and 2. Here the loop variable `index` is a **position**; `numbers[index]` is the value stored there. Assigning to `numbers[index]` changes the list.

````python
numbers = [3, 5, 2]

for index in range(len(numbers)):
    numbers[index] = numbers[index] * 2

print(numbers)
````

**Output**
```text
[6, 10, 4]
```

## 12.9 Use a while loop to search a list

### Example: Visit a list using an index and while

A `while` loop repeats while its condition is true. Unlike `for`, it does not automatically select the next element. We initialise `index`, check it against the list length and increase it ourselves.

````python
numbers = [3, 5, 7]
index = 0

while index < len(numbers):
    print(numbers[index])
    index += 1
````

**Output**
```text
3
5
7
```

### Example: Find the first temperature above 25

Stop as soon as a high temperature is found. `break` leaves the loop immediately. `found` records whether a match was found; an empty list or a list with no match leaves it `False`.

````python
temperatures = [18, 21, 27, 24]
index = 0
found = False

while index < len(temperatures):
    if temperatures[index] > 25:
        found = True
        break
    index += 1

if found:
    print("Position:", index + 1)
    print("Temperature:", temperatures[index])
else:
    print("No high temperature")
````

**Output**
```text
Position: 3
Temperature: 27
```

### Example: Build a list when the number of readings is unknown

## 12.11 Build an arithmetic sequence

A **sequence** is an ordered collection of terms. In an arithmetic sequence, we add the same amount each time: for example, 3, 6, 9, 12, … .

### Example: Make a list of multiples of 3

`range(3, 16, 3)` supplies 3, 6, 9, 12 and 15. The stop value is excluded. Append each number to store the sequence in a list.

````python
multiples = []

for number in range(3, 16, 3):
    multiples.append(number)

print(multiples)
````

## 12.12 Build the Fibonacci sequence

### Example: Generate eight Fibonacci numbers

Start with 0 and 1. Each new term is the sum of the previous two. Negative indices `[-1]` and `[-2]` select the last two stored terms. Two terms already exist, so six loop passes produce eight terms in total.

````python
fibonacci = [0, 1]
number_of_terms = 8

for term_number in range(2, number_of_terms):
    next_term = fibonacci[-1] + fibonacci[-2]
    fibonacci.append(next_term)

print(fibonacci)
````

**Output**
```text
[0, 1, 1, 2, 3, 5, 8, 13]
```

For the first new term, `0 + 1 = 1`; next, `1 + 1 = 2`; then `1 + 2 = 3`. After every append, the last two elements have changed. `term_number` controls how many passes run; it is not itself a Fibonacci number.

## 12.13 Build a geometric sequence

### Example: Model a signal that halves at each step

In a geometric sequence, multiply by the same factor to get the next term. This example starts at 80 and multiplies by 0.5. Append the current value **before** updating it, so the initial value is included.

````python
signal = 80.0
samples = []

for step in range(5):
    samples.append(signal)
    signal = signal * 0.5

print(samples)
````

## 12.14 Add the terms of a series

A **sequence** lists terms; a **series** adds them. For example, the sequence 1, 1/2, 1/3 has the series total 1 + 1/2 + 1/3.

### Example: Calculate a harmonic-series partial sum

A partial sum adds a finite number of terms. Here term number `n` gives the value `1 / n`. Store each term and add it to the total. Start at 1 to avoid division by zero.

````python
number_of_terms = 4
terms = []
total = 0

for n in range(1, number_of_terms + 1):
    term = 1 / n
    terms.append(term)
    total += term

print("Terms:", terms)
print("Total:", round(total, 4))
````

**Output**
```text
Terms: [1.0, 0.5, 0.3333333333333333, 0.25]
Total: 2.0833
```

`round(total, 4)` rounds the displayed result to four decimal places; the running total is calculated without rounding.

## 12.15 Add a series with alternating signs

### Example: Calculate 1 − 1/2 + 1/3 − 1/4 + …

Use a separate `sign` variable. It starts at 1 and is multiplied by -1 after each term, so its values alternate between 1 and -1.

````python
number_of_terms = 4
sign = 1
terms = []
total = 0

for n in range(1, number_of_terms + 1):
    term = sign / n
    terms.append(term)
    total += term
    sign = -sign

print("Terms:", terms)
print("Total:", round(total, 4))
````

**Output**
```text
Terms: [1.0, -0.5, 0.3333333333333333, -0.25]
Total: 0.5833
```

## 12.16 Find prime numbers and store them in a list

### Example: Test whether one number is prime

A prime is an integer greater than 1 with no divisors other than 1 and itself. `%` gives the remainder: `number % divisor == 0` means exact division. Begin with `True` for a number above 1; finding any divisor changes the result to `False`.

````python
number = 15
is_prime = number > 1

for divisor in range(2, number):
    if number % divisor == 0:
        is_prime = False
        break

print("Prime:", is_prime)
````

**Output**
```text
Prime: False
```

Try 13, 2 and 1 as well. For 2 there are no divisors to test, so it remains prime. For 1 the flag starts `False`. Once a divisor is found, `break` ends the search.

### Example: Build a list of all primes from 2 to 20

Now repeat the prime test for each candidate. The **outer loop** chooses a candidate; the **inner loop** tests its divisors. Reset `is_prime` for every candidate, then append the candidate only if it passes. `break` leaves only the inner loop.

````python
primes = []

for number in range(2, 21):
    is_prime = True
    for divisor in range(2, number):
        if number % divisor == 0:
            is_prime = False
            break
    if is_prime:
        primes.append(number)

print(primes)
````

**Output**
```text
[2, 3, 5, 7, 11, 13, 17, 19]
```
